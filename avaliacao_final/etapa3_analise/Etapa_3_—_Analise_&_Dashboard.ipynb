```python
# Célula 3 — Leitura das tabelas GOLD

gold_channel = pd.read_parquet(PATH_CHANNEL)
gold_daily = pd.read_parquet(PATH_DAILY)
gold_category = pd.read_parquet(PATH_CATEGORY)
gold_risk = pd.read_parquet(PATH_RISK)

# Garantir tipo correto da data
gold_daily["date"] = pd.to_datetime(gold_daily["date"])

print("✓ Tabelas GOLD carregadas")
print()

print(f"gold_channel  : {len(gold_channel):,} linhas")
print(f"gold_daily    : {len(gold_daily):,} linhas")
print(f"gold_category : {len(gold_category):,} linhas")
print(f"gold_risk     : {len(gold_risk):,} linhas")
```

```python
# Célula 5 — Preparação dos dados

# Cópias para não alterar as tabelas Gold originais
df_channel = gold_channel.copy()
df_daily = gold_daily.copy()
df_category = gold_category.copy()
df_risk = gold_risk.copy()

# Normalização dos textos
df_channel["channel"] = df_channel["channel"].astype(str).str.lower()
df_risk["risk_level"] = df_risk["risk_level"].astype(str)

# Ordenação temporal
df_daily = df_daily.sort_values("date")

# Conversão das taxas para percentual
df_channel["fraud_rate_pct"] = df_channel["fraud_rate"] * 100
df_category["fraud_rate_pct"] = df_category["fraud_rate"] * 100
df_risk["fraud_rate_pct"] = df_risk["fraud_rate"] * 100
df_daily["fraud_rate_pct"] = df_daily["fraud_rate"] * 100

print("✓ Dados preparados")
```

```python
# Célula 6 — KPIs

total_transactions = int(df_channel["total_transactions"].sum())
total_fraud = int(df_channel["fraud_transactions"].sum())
total_fraud_amount = float(df_channel["fraud_amount"].sum())

fraud_rate = (
    total_fraud / total_transactions
    if total_transactions > 0
    else 0
)

print("KPIs GERAIS")
print("-" * 40)
print(f"Transações : {total_transactions:,}")
print(f"Fraudes    : {total_fraud:,}")
print(f"Taxa       : {fraud_rate:.2%}")
print(f"Valor      : R$ {total_fraud_amount:,.2f}")
```

```python
# Célula 7 — Gráficos

# ------------------------------------------------------------
# TENDÊNCIA
# ------------------------------------------------------------

fig_daily = px.line(
    df_daily,
    x="date",
    y="fraud_rate_pct",
    markers=True,
    title="Evolução diária da taxa de fraude",
    labels={
        "date": "Data",
        "fraud_rate_pct": "Taxa de fraude (%)"
    }
)

fig_daily.update_layout(
    template="plotly_white",
    hovermode="x unified",
    margin=dict(l=50, r=30, t=60, b=50)
)

# ------------------------------------------------------------
# COMPOSIÇÃO POR CANAL
# ------------------------------------------------------------

fig_channel = px.bar(
    df_channel.sort_values("fraud_transactions", ascending=False),
    x="channel",
    y="fraud_transactions",
    text="fraud_transactions",
    title="Fraudes por canal",
    labels={
        "channel": "Canal",
        "fraud_transactions": "Fraudes"
    }
)

fig_channel.update_traces(
    textposition="outside"
)

fig_channel.update_layout(
    template="plotly_white",
    margin=dict(l=50, r=30, t=60, b=50)
)

# ------------------------------------------------------------
# COMPOSIÇÃO POR NÍVEL DE RISCO
# ------------------------------------------------------------

fig_risk = px.bar(
    df_risk.sort_values("fraud_rate_pct", ascending=False),
    x="risk_level",
    y="fraud_rate_pct",
    text="fraud_rate_pct",
    title="Taxa de fraude por nível de risco",
    labels={
        "risk_level": "Nível de risco",
        "fraud_rate_pct": "Taxa de fraude (%)"
    }
)

fig_risk.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig_risk.update_layout(
    template="plotly_white",
    margin=dict(l=50, r=30, t=60, b=50)
)

print("✓ Gráficos criados")
```

```python
# Célula 8 — Dados para o dashboard HTML

def dataframe_to_records(df):
    """Converte DataFrame para registros JSON serializáveis."""
    temp = df.copy()

    for col in temp.columns:
        if pd.api.types.is_datetime64_any_dtype(temp[col]):
            temp[col] = temp[col].dt.strftime("%Y-%m-%d")

    temp = temp.replace({np.nan: None})

    return temp.to_dict(orient="records")


dashboard_data = {
    "channel": dataframe_to_records(df_channel),
    "daily": dataframe_to_records(df_daily),
    "category": dataframe_to_records(df_category),
    "risk": dataframe_to_records(df_risk),
}

dashboard_json = json.dumps(
    dashboard_data,
    ensure_ascii=False
)

print("✓ Dados preparados para o HTML")
print(f"Tamanho dos dados: {len(dashboard_json):,} caracteres")
```

```python
# Célula 9 — Geração do Dashboard HTML

OUTPUT_DIR = "../etapa3_analise/"

os.makedirs(OUTPUT_DIR, exist_ok=True)

OUTPUT_HTML = os.path.join(
    OUTPUT_DIR,
    "dashboard_techpay.html"
)

html = f"""
<!DOCTYPE html>

<html lang="pt-BR">

<head>

<meta charset="UTF-8">

<meta name="viewport"
      content="width=device-width, initial-scale=1.0">

<title>TECHPAY — Dashboard de Risco</title>

<script src="https://cdn.plot.ly/plotly-2.35.2.min.js"></script>

<style>

* {{
    box-sizing: border-box;
}}

body {{
    margin: 0;
    padding: 0;

    font-family:
        Arial,
        Helvetica,
        sans-serif;

    background: #f4f6f8;
    color: #222;
}}

.container {{
    max-width: 1400px;
    margin: 0 auto;
    padding: 25px;
}}

.header {{
    background: white;
    border-radius: 12px;
    padding: 25px;
    margin-bottom: 20px;

    box-shadow:
        0 2px 8px rgba(0,0,0,0.08);
}}

.header h1 {{
    margin: 0 0 8px 0;
    font-size: 30px;
}}

.header p {{
    margin: 0;
    color: #666;
}}

.filters {{
    display: flex;
    gap: 20px;
    flex-wrap: wrap;

    background: white;
    padding: 20px;
    margin-bottom: 20px;

    border-radius: 12px;

    box-shadow:
        0 2px 8px rgba(0,0,0,0.08);
}}

.filter {{
    display: flex;
    flex-direction: column;
    gap: 7px;

    min-width: 200px;
}}

.filter label {{
    font-weight: bold;
    font-size: 14px;
}}

.filter select {{
    padding: 10px;
    border-radius: 6px;
    border: 1px solid #ccc;
    background: white;
    font-size: 14px;
}}

.kpis {{
    display: grid;

    grid-template-columns:
        repeat(4, 1fr);

    gap: 20px;

    margin-bottom: 20px;
}}

.kpi {{
    background: white;
    padding: 22px;

    border-radius: 12px;

    box-shadow:
        0 2px 8px rgba(0,0,0,0.08);
}}

.kpi-title {{
    color: #666;
    font-size: 14px;
    margin-bottom: 10px;
}}

.kpi-value {{
    font-size: 28px;
    font-weight: bold;
}}

.section {{
    background: white;
    border-radius: 12px;
    padding: 20px;
    margin-bottom: 20px;

    box-shadow:
        0 2px 8px rgba(0,0,0,0.08);
}}

.section h2 {{
    margin-top: 0;
    font-size: 20px;
}}

.two-columns {{
    display: grid;

    grid-template-columns:
        repeat(2, 1fr);

    gap: 20px;
}}

.chart {{
    width: 100%;
    min-height: 400px;
}}

table {{
    width: 100%;
    border-collapse: collapse;
}}

th, td {{
    padding: 12px;
    text-align: left;

    border-bottom:
        1px solid #eee;
}}

th {{
    background: #f7f7f7;
}}

tbody tr:hover {{
    background: #fafafa;
}}

.badge {{
    display: inline-block;
    padding: 5px 9px;
    border-radius: 20px;
    font-size: 12px;
    font-weight: bold;
}}

.footer {{
    text-align: center;
    color: #777;
    font-size: 13px;
    padding: 20px;
}}

@media (max-width: 900px) {{

    .kpis {{
        grid-template-columns:
            repeat(2, 1fr);
    }}

    .two-columns {{
        grid-template-columns: 1fr;
    }}

}}

@media (max-width: 600px) {{

    .kpis {{
        grid-template-columns: 1fr;
    }}

}}

</style>

</head>


<body>

<div class="container">

    <!-- ================================================= -->
    <!-- CABEÇALHO -->
    <!-- ================================================= -->

    <div class="header">

        <h1>TECHPAY — Dashboard de Risco</h1>

        <p>
            Monitoramento de transações, fraudes e níveis de risco
        </p>

    </div>


    <!-- ================================================= -->
    <!-- FILTROS -->
    <!-- ================================================= -->

    <div class="filters">

        <div class="filter">

            <label for="channelFilter">
                Canal
            </label>

            <select id="channelFilter"
                    onchange="updateDashboard()">

                <option value="ALL">
                    Todos
                </option>

            </select>

        </div>


        <div class="filter">

            <label for="riskFilter">
                Nível de risco
            </label>

            <select id="riskFilter"
                    onchange="updateDashboard()">

                <option value="ALL">
                    Todos
                </option>

            </select>

        </div>

    </div>


    <!-- ================================================= -->
    <!-- KPIs -->
    <!-- ================================================= -->

    <div class="kpis">

        <div class="kpi">

            <div class="kpi-title">
                Total de transações
            </div>

            <div class="kpi-value"
                 id="kpiTransactions">
                -
            </div>

        </div>


        <div class="kpi">

            <div class="kpi-title">
                Transações fraudulentas
            </div>

            <div class="kpi-value"
                 id="kpiFraud">
                -
            </div>

        </div>


        <div class="kpi">

            <div class="kpi-title">
                Taxa de fraude
            </div>

            <div class="kpi-value"
                 id="kpiFraudRate">
                -
            </div>

        </div>


        <div class="kpi">

            <div class="kpi-title">
                Valor fraudado
            </div>

            <div class="kpi-value"
                 id="kpiFraudAmount">
                -
            </div>

        </div>

    </div>


    <!-- ================================================= -->
    <!-- TENDÊNCIA -->
    <!-- ================================================= -->

    <div class="section">

        <h2>
            Tendência
        </h2>

        <div id="dailyChart"
             class="chart">
        </div>

    </div>


    <!-- ================================================= -->
    <!-- COMPOSIÇÃO -->
    <!-- ================================================= -->

    <div class="two-columns">

        <div class="section">

            <h2>
                Composição por canal
            </h2>

            <div id="channelChart"
                 class="chart">
            </div>

        </div>


        <div class="section">

            <h2>
                Composição por nível de risco
            </h2>

            <div id="riskChart"
                 class="chart">
            </div>

        </div>

    </div>


    <!-- ================================================= -->
    <!-- DETALHE -->
    <!-- ================================================= -->

    <div class="section">

        <h2>
            Detalhamento por categoria
        </h2>

        <div style="overflow-x:auto;">

            <table>

                <thead>

                    <tr>

                        <th>
                            Categoria
                        </th>

                        <th>
                            Transações
                        </th>

                        <th>
                            Fraudes
                        </th>

                        <th>
                            Taxa de fraude
                        </th>

                        <th>
                            Valor fraudado
                        </th>

                    </tr>

                </thead>

                <tbody id="categoryTable">

                </tbody>

            </table>

        </div>

    </div>


    <div class="footer">

        TECHPAY — Projeto de Big Data

    </div>

</div>


<script>

const DATA = {dashboard_json};


// ============================================================
// FORMATAÇÃO
// ============================================================

function formatNumber(value) {{

    return new Intl.NumberFormat(
        'pt-BR'
    ).format(value);

}}


function formatCurrency(value) {{

    return new Intl.NumberFormat(
        'pt-BR',
        {{
            style: 'currency',
            currency: 'BRL'
        }}
    ).format(value);

}}


function formatPercent(value) {{

    return new Intl.NumberFormat(
        'pt-BR',
        {{
            minimumFractionDigits: 2,
            maximumFractionDigits: 2
        }}
    ).format(value) + '%';

}}


// ============================================================
// FILTROS
// ============================================================

function populateFilters() {{

    const channelFilter =
        document.getElementById(
            'channelFilter'
        );

    const riskFilter =
        document.getElementById(
            'riskFilter'
        );


    const channels = [
        ...new Set(
            DATA.channel.map(
                row => row.channel
            )
        )
    ].sort();


    const risks = [
        ...new Set(
            DATA.risk.map(
                row => row.risk_level
            )
        )
    ].sort();


    channels.forEach(channel => {{

        const option =
            document.createElement(
                'option'
            );

        option.value = channel;
        option.textContent =
            channel.toUpperCase();

        channelFilter.appendChild(
            option
        );

    }});


    risks.forEach(risk => {{

        const option =
            document.createElement(
                'option'
            );

        option.value = risk;
        option.textContent = risk;

        riskFilter.appendChild(
            option
        );

    }});

}}


// ============================================================
// KPIs
// ============================================================

function updateKPIs(
    channelFilter,
    riskFilter
) {{

    let channelData =
        DATA.channel;


    // --------------------------------------------------------
    // Filtro por canal
    // --------------------------------------------------------

    if (channelFilter !== 'ALL') {{

        channelData =
            channelData.filter(
                row =>
                    row.channel === channelFilter
            );

    }}


    // --------------------------------------------------------
    // Cálculo dos KPIs
    // --------------------------------------------------------

    let transactions = 0;
    let fraud = 0;
    let fraudAmount = 0;


    channelData.forEach(row => {{

        transactions +=
            Number(row.total_transactions || 0);

        fraud +=
            Number(row.fraud_transactions || 0);

        fraudAmount +=
            Number(row.fraud_amount || 0);

    }});


    // --------------------------------------------------------
    // Quando existe filtro de risco,
    // usamos a tabela risk para ajustar a análise.
    // --------------------------------------------------------

    if (riskFilter !== 'ALL') {{

        const riskData =
            DATA.risk.filter(
                row =>
                    row.risk_level === riskFilter
            );


        transactions =
            riskData.reduce(
                (sum, row) =>
                    sum +
                    Number(
                        row.total_transactions || 0
                    ),
                0
            );


        fraud =
            riskData.reduce(
                (sum, row) =>
                    sum +
                    Number(
                        row.fraud_transactions || 0
                    ),
                0
            );


        fraudAmount =
            riskData.reduce(
                (sum, row) =>
                    sum +
                    Number(
                        row.fraud_amount || 0
                    ),
                0
            );

    }}


    const rate =
        transactions > 0
            ? (fraud / transactions) * 100
            : 0;


    document.getElementById(
        'kpiTransactions'
    ).textContent =
        formatNumber(transactions);


    document.getElementById(
        'kpiFraud'
    ).textContent =
        formatNumber(fraud);


    document.getElementById(
        'kpiFraudRate'
    ).textContent =
        formatPercent(rate);


    document.getElementById(
        'kpiFraudAmount'
    ).textContent =
        formatCurrency(fraudAmount);

}}


// ============================================================
// GRÁFICO DE TENDÊNCIA
// ============================================================

function updateDailyChart() {{

    const dates =
        DATA.daily.map(
            row => row.date
        );


    const rates =
        DATA.daily.map(
            row =>
                Number(
                    row.fraud_rate_pct || 0
                )
        );


    const fraud =
        DATA.daily.map(
            row =>
                Number(
                    row.fraud_transactions || 0
                )
        );


    const trace = {{

        x: dates,

        y: rates,

        customdata: fraud,

        mode: 'lines+markers',

        type: 'scatter',

        hovertemplate:
            '<b>%{{x}}</b>' +
            '<br>Taxa: %{{y:.2f}}%' +
            '<br>Fraudes: %{{customdata}}' +
            '<extra></extra>'

    }};


    const layout = {{

        margin:
        {{
            l: 55,
            r: 30,
            t: 20,
            b: 50
        }},

        xaxis:
        {{
            title: 'Data'
        }},

        yaxis:
        {{
            title: 'Taxa de fraude (%)'
        }},

        hovermode:
            'x unified',

        template:
            'plotly_white'

    }};


    Plotly.react(
        'dailyChart',
        [trace],
        layout,
        {{responsive: true}}
    );

}}


// ============================================================
// GRÁFICO POR CANAL
// ============================================================

function updateChannelChart(
    channelFilter
) {{

    let data =
        DATA.channel;


    if (channelFilter !== 'ALL') {{

        data =
            data.filter(
                row =>
                    row.channel === channelFilter
            );

    }}


    const channels =
        data.map(
            row => row.channel
        );


    const fraud =
        data.map(
            row =>
                Number(
                    row.fraud_transactions || 0
                )
        );


    const trace = {{

        x: channels,

        y: fraud,

        text: fraud,

        textposition:
            'outside',

        type: 'bar',

        hovertemplate:
            '<b>%{{x}}</b>' +
            '<br>Fraudes: %{{y}}' +
            '<extra></extra>'

    }};


    const layout = {{

        margin:
        {{
            l: 50,
            r: 30,
            t: 20,
            b: 50
        }},

        xaxis:
        {{
            title: 'Canal'
        }},

        yaxis:
        {{
            title: 'Fraudes'
        }},

        template:
            'plotly_white'

    }};


    Plotly.react(
        'channelChart',
        [trace],
        layout,
        {{responsive: true}}
    );

}}


// ============================================================
// GRÁFICO POR RISCO
// ============================================================

function updateRiskChart(
    riskFilter
) {{

    let data =
        DATA.risk;


    if (riskFilter !== 'ALL') {{

        data =
            data.filter(
                row =>
                    row.risk_level === riskFilter
            );

    }}


    const risks =
        data.map(
            row =>
                row.risk_level
        );


    const rates =
        data.map(
            row =>
                Number(
                    row.fraud_rate_pct || 0
                )
        );


    const trace = {{

        x: risks,

        y: rates,

        text: rates,

        textposition:
            'outside',

        type: 'bar',

        hovertemplate:
            '<b>%{{x}}</b>' +
            '<br>Taxa: %{{y:.2f}}%' +
            '<extra></extra>'

    }};


    const layout = {{

        margin:
        {{
            l: 50,
            r: 30,
            t: 20,
            b: 50
        }},

        xaxis:
        {{
            title: 'Nível de risco'
        }},

        yaxis:
        {{
            title: 'Taxa de fraude (%)'
        }},

        template:
            'plotly_white'

    }};


    Plotly.react(
        'riskChart',
        [trace],
        layout,
        {{responsive: true}}
    );

}}


// ============================================================
// TABELA DE DETALHES
// ============================================================

function updateCategoryTable() {{

    const tbody =
        document.getElementById(
            'categoryTable'
        );


    tbody.innerHTML = '';


    const data =
        [...DATA.category]
        .sort(
            (a, b) =>
                Number(b.fraud_rate_pct || 0)
                -
                Number(a.fraud_rate_pct || 0)
        );


    data.forEach(row => {{

        const tr =
            document.createElement('tr');


        const category =
            document.createElement('td');

        category.textContent =
            row.merchant_category;


        const transactions =
            document.createElement('td');

        transactions.textContent =
            formatNumber(
                Number(
                    row.total_transactions || 0
                )
            );


        const fraud =
            document.createElement('td');

        fraud.textContent =
            formatNumber(
                Number(
                    row.fraud_transactions || 0
                )
            );


        const rate =
            document.createElement('td');

        rate.textContent =
            formatPercent(
                Number(
                    row.fraud_rate_pct || 0
                )
            );


        const amount =
            document.createElement('td');

        amount.textContent =
            formatCurrency(
                Number(
                    row.fraud_amount || 0
                )
            );


        tr.appendChild(category);
        tr.appendChild(transactions);
        tr.appendChild(fraud);
        tr.appendChild(rate);
        tr.appendChild(amount);


        tbody.appendChild(tr);

    }});

}}


// ============================================================
// ATUALIZAÇÃO GERAL
// ============================================================

function updateDashboard() {{

    const channelFilter =
        document.getElementById(
            'channelFilter'
        ).value;


    const riskFilter =
        document.getElementById(
            'riskFilter'
        ).value;


    updateKPIs(
        channelFilter,
        riskFilter
    );


    updateDailyChart();


    updateChannelChart(
        channelFilter
    );


    updateRiskChart(
        riskFilter
    );


    updateCategoryTable();

}}


// ============================================================
// INICIALIZAÇÃO
// ============================================================

populateFilters();

updateDashboard();

</script>

</body>

</html>
"""


with open(
    OUTPUT_HTML,
    "w",
    encoding="utf-8"
) as f:

    f.write(html)


print("=" * 70)
print("✓ DASHBOARD GERADO COM SUCESSO")
print("=" * 70)
print()
print(f"Arquivo:")
print(OUTPUT_HTML)
print()
print(f"Tamanho: {os.path.getsize(OUTPUT_HTML) / 1024:.1f} KB")
```
